# Rectified Linear Unit (ReLU)
**The most popular activation function in Deep Learning.**

## 1. What is ReLU?
The **Rectified Linear Unit (ReLU)** is a simple, non-linear activation function used in neural networks. 
It allows models to learn complex patterns and solves the "vanishing gradient" problem found in older functions like Sigmoid.

**Formula:**
$$f(x) = \max(0, x)$$

**Properties:**
* **Output:** $x$ if $x > 0$, else $0$.
* **Range:** $[0, \infty)$
* **Speed:** Very fast to compute (just a threshold check).

```python
import numpy as np
import matplotlib.pyplot as plt

# Check for optional Deep Learning libraries
try:
    import torch
    HAS_TORCH = True
except ImportError:
    HAS_TORCH = False

try:
    import tensorflow as tf
    HAS_TF = True
except ImportError:
    HAS_TF = False
    
print(f"NumPy version: {np.__version__}")
print(f"PyTorch available: {HAS_TORCH}")
print(f"TensorFlow available: {HAS_TF}")
```

## 2. Implementation from Scratch
We use `numpy` to implement ReLU efficiently for arrays. We also implement its derivative, which is needed for backpropagation.

* **Derivative:** $1$ if $x > 0$, else $0$.

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Generate data
x = np.linspace(-10, 10, 400)
y_relu = relu(x)
y_grad = relu_derivative(x)
```

## 3. Visualization
Here we plot ReLU (Blue) and its gradient (Orange). Note how the gradient is exactly 0 for negative inputs, which creates "sparsity" in the network.

```python
plt.figure(figsize=(10, 6))

# Plot ReLU
plt.plot(x, y_relu, label='ReLU Function $f(x) = max(0, x)$', linewidth=3, color='#1f77b4')

# Plot Derivative (dashed)
plt.plot(x, y_grad, label='Gradient $f\'(x)$', linestyle='--', linewidth=2, color='#ff7f0e')

# Styling
plt.title('ReLU Activation and its Derivative', fontsize=16)
plt.axhline(0, color='black', linewidth=1, alpha=0.5)
plt.axvline(0, color='black', linewidth=1, alpha=0.5)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.xlabel('Input (x)', fontsize=12)
plt.ylabel('Output', fontsize=12)
plt.show()
```

## 4. Comparison with Other Functions
* **Sigmoid:** Squashes output between 0 and 1. Good for probabilities, but suffers from vanishing gradients.
* **Leaky ReLU:** Fixes the "Dying ReLU" problem by allowing a small gradient for negative inputs.

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def leaky_relu(x, alpha=0.1):
    return np.where(x > 0, x, alpha * x)

plt.figure(figsize=(10, 6))
plt.plot(x, relu(x), label='ReLU', linewidth=2)
plt.plot(x, sigmoid(x), label='Sigmoid', linestyle=':', linewidth=2)
plt.plot(x, leaky_relu(x), label='Leaky ReLU (alpha=0.1)', linestyle='-.', linewidth=2)

plt.title('Comparison of Activation Functions', fontsize=16)
plt.ylim(-2, 5)  # Zoom in to see the differences
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 5. Using ReLU in Deep Learning Frameworks
If you are using PyTorch or TensorFlow, you don't need to write the math yourself.

```python
# PyTorch Example
if HAS_TORCH:
    print("--- PyTorch ReLU ---")
    tensor_x = torch.tensor([-2.0, 0.0, 2.0])
    relu_layer = torch.nn.ReLU()
    print(f"Input: {tensor_x}")
    print(f"Output: {relu_layer(tensor_x)}")
else:
    print("PyTorch not installed.")

# TensorFlow Example
if HAS_TF:
    print("\n--- TensorFlow ReLU ---")
    tf_x = tf.constant([-2.0, 0.0, 2.0])
    tf_out = tf.nn.relu(tf_x)
    print(f"Input: {tf_x.numpy()}")
    print(f"Output: {tf_out.numpy()}")
else:
    print("\nTensorFlow not installed.")
```
